```python
import pandas as pd
import re
from collections import Counter

# 1. Загружаем файл (обрати внимание на разделитель ';')
df = pd.read_csv('Messages_Arknights [English Server].csv', sep=';')

# Очистим данные от пустых строк в тексте (если есть)
df = df.dropna(subset=['message'])

print("=== ОБЩАЯ СТАТИСТИКА ===")
print(f"Всего сообщений в файле: {len(df)}")

# 2. Самый просматриваемый пост
top_view = df.loc[df['views'].idxmax()]
print(f"\nПост с макс. просмотрами ({top_view['views']}):")
print(f"Дата: {top_view['date']}")
print(f"Текст (первые 100 символов): {top_view['message'][:100]}...")

# 3. Самый комментируемый пост
top_comment = df.loc[df['comments'].idxmax()]
print(f"\nПост с макс. комментариями ({top_comment['comments']}):")
print(f"Текст (первые 100 символов): {top_comment['message'][:100]}...")

# 4. Самые частые слова (исключаем мелкие слова и ссылки)
print("\n=== ТОП-10 САМЫХ ЧАСТЫХ СЛОВ ===")
all_text = " ".join(df['message'].astype(str))
# Убираем ссылки и знаки препинания
all_text = re.sub(r'http\S+', '', all_text)
all_text = re.sub(r'[^\w\s]', '', all_text)
words = all_text.lower().split()
# Исключаем предлоги и союзы (очень короткие слова)
filtered_words = [w for w in words if len(w) > 3]
top_words = Counter(filtered_words).most_common(10)
for word, count in top_words:
    print(f"{word}: {count} раз")

# 5. Средняя длина поста
df['text_length'] = df['message'].astype(str).apply(len)
print(f"\n=== ДЛИНА ПОСТОВ ===")
print(f"Средняя длина поста: {df['text_length'].mean():.0f} символов")
print(f"Самый длинный пост: {df['text_length'].max()} символов")
```

```python
import pandas as pd

# Загружаем наш исходный файл
df = pd.read_csv('Messages_Arknights [English Server].csv', sep=';')

# Берем только нужные столбцы для красивой таблицы
# (Дата, ID сообщения, Просмотры, Реакции, Текст)
table = df[['date', 'message_id', 'views', 'reactions_total', 'message']].copy()

# Сортируем по просмотрам (от большего к меньшему)
table = table.sort_values(by='views', ascending=False)

# Берем топ-10 постов
top_10 = table.head(10)

# Выводим на экран
print("ТОП-10 САМЫХ ПРОСМАТРИВАЕМЫХ ПОСТОВ:")
display(top_10) # В Colab display выводит красивую таблицу, а не просто текст

# Сохраняем эту таблицу в новый файл
top_10.to_csv('top_10_posts.csv', index=False, sep=';')
print("\nТаблица сохранена в файл 'top_10_posts.csv'")
```
