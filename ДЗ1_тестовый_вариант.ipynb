```python
import requests
from bs4 import BeautifulSoup

# 1. Отправляем запрос на сайт (URL)
url = "https://books.toscrape.com/"
response = requests.get(url)

# 2. Проверяем, что сайт ответил (код 200 значит "Успех")
print("Статус ответа:", response.status_code)

# 3. Передаем код страницы в BeautifulSoup для разбора
soup = BeautifulSoup(response.text, 'html.parser')

# 4. Ищем все элементы h3 (в них лежат названия книг)
books = soup.find_all('h3')

# 5. Выводим первые 5 названий на экран
print("Первые 5 книг на сайте:")
for i, book in enumerate(books[:5], 1):
    title = book.get_text(strip=True)
    print(f"{i}. {title}")
```
